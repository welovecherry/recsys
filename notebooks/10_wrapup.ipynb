### <mark style="display:block; background:#f8bbd0; color:#1a1a1a; padding:6px 12px; border-radius:4px">🎯 10.1 모델 비교표 — 네 가지 추천 방식을 같은 기준으로 채점한다</mark>

**흐름에서의 위치:** 지난 주차의 모델 → **같은 데이터 · 같은 채점**으로 다시 계산 → 한 표.

**10.1 에서 하는 일** — 1 · 2 · 4 · 6주차의 추천 방식을 **함수 네 개**로 만들고, 각각 Recall@10 · 성향에 맞지 않는 추천 수 · 만드는 데 걸린 시간을 한 표로 모읍니다.

**왜 하나**
- 열 주 동안 점수는 주차마다 따로 나왔습니다. 한 표에 놓아야 **무엇을 얻고 무엇이 낮아졌는지** 보입니다.
- 모두 **같은 시간순 분할**(학습 구간 7,095건 · 채점 구간 991건)로 채점하므로 공정한 비교입니다.

**네 가지 방식**

| 주차 | 방식 | 하는 일 |
|---|---|---|
| 1 | 전체 인기순 | 모두에게 가장 많이 담긴 종목 10개를 똑같이 추천 |
| 2 | 세그먼트별 인기순 | 취향이 비슷한 무리를 나누고 **무리의 인기 종목**을 추천 (기록이 없는 신규 투자자는 전체 인기순) |
| 4 | 행렬분해 (상한 없음) | 사용자 요인 × 아이템 요인으로 계산한 점수 순서대로 추천 |
| 6 | 행렬분해 + 위험도 상한 | 4주차와 같되, 투자 성향이 허용하는 위험도(안정형 3 · 성장형 4 · 공격형 5)를 넘는 종목을 거른다 |

**성향에 맞지 않는 추천** — 300명에게 추천한 목록에서 그 투자자의 위험도 상한을 넘는 종목의 개수입니다 (6주차 6.9 와 같은 방법).

```python
# 이 셀에서 하는 일 — 추천 방식 네 가지를 함수로 만든다 (1주차 · 2주차 · 4주차 · 6주차)
# 왜 하나 — 같은 모양의 함수로 만들어야 같은 채점 함수에 넣어 공정하게 비교할 수 있다

import time                                                                  # 걸린 시간을 재는 도구
import pandas as pd                                                          # 표를 다루는 도구
import recsys                                                                # 지난 주차에서 쓴 도구 모음
import recommend_batch                                                       # 7주차 스크립트 (학습하기 · 추천하기)

items, users, interactions = recsys.load()
train, test, 기준시점 = recsys.split_by_time(interactions)                    # 3주차부터 쓰던 시간순 분할

성향 = users.set_index("user_id")["persona"].to_dict()                        # {투자자 번호: 투자 성향}
위험도 = items.set_index("item_id")["risk_level"].to_dict()                   # {종목 번호: 위험도}
상한 = recommend_batch.상한                                                    # {"안정형": 3, "성장형": 4, "공격형": 5}
이미_담은것 = train.groupby("user_id")["item_id"].apply(set).to_dict()         # {투자자: 학습 구간에서 이미 담은 종목}

걸린_시간 = {}                                                                # {방식 이름: 만드는 데 걸린 ms}

# ── 1주차 · 전체 인기순 ──
시작 = time.perf_counter()
인기순위 = list(train["item_id"].value_counts().index)                        # 많이 담긴 순서의 종목 번호
걸린_시간["1주차"] = (time.perf_counter() - 시작) * 1000

def 인기순_추천(사람, 이미):
    return recsys.take(인기순위, 이미)                                         # 이미 담은 것을 빼고 위에서 10개

# ── 2주차 · 세그먼트별 인기순 ──
시작 = time.perf_counter()
세그먼트, 세그먼트별_순위 = recsys.segments(train, items)                      # 취향이 비슷한 무리와 무리별 인기 순위
걸린_시간["2주차"] = (time.perf_counter() - 시작) * 1000

def 세그먼트_추천(사람, 이미):
    if 사람 in 세그먼트:
        return recsys.take(세그먼트별_순위[세그먼트[사람]], 이미)
    return recsys.take(인기순위, 이미)                                         # 학습 구간에 기록이 없는 신규 투자자

# ── 4주차 · 6주차 · 행렬분해 (같은 모델에서 상한만 다르다) ──
시작 = time.perf_counter()
학습_결과 = recommend_batch.학습하기(train, items)                             # 행렬분해로 사용자 요인 · 아이템 요인을 만든다
걸린_시간["4주차 · 6주차"] = (time.perf_counter() - 시작) * 1000

상한_끔 = {번호: 1 for 번호 in 위험도}                                          # 모든 종목의 위험도를 1 로 두면 상한에 걸리는 종목이 없다

def 행렬분해_추천(사람, 이미):                                                  # 4주차 — 상한 없음
    return [번호 for 번호, 점수 in recommend_batch.추천하기(학습_결과, 사람, 성향, 상한_끔)]

def 상한_추천(사람, 이미):                                                      # 6주차 — 위험도 상한 적용
    return [번호 for 번호, 점수 in recommend_batch.추천하기(학습_결과, 사람, 성향, 위험도)]

print("추천 방식 4개를 만들었습니다. 학습 구간", f"{len(train):,}건 · 채점 구간 {len(test):,}건")
```

**결과 읽는 법**

**결론: 추천 방식 네 가지가 같은 모양의 함수가 되었고, 같은 학습 구간(7,095건)으로 만들었습니다.**
- 출력은 학습 구간과 채점 구간의 기록 수입니다. 분할 기준은 3주차와 같은 시간순 분할입니다
- 4주차와 6주차는 **같은 모델**입니다. 6주차 함수만 위험도 상한을 켠 것입니다. 4주차 함수는 모든 종목의 위험도를 1 로 두어 상한을 끕니다

```python
# 이 셀에서 하는 일 — 네 가지 방식의 점수 · 성향에 맞지 않는 추천 수 · 걸린 시간을 한 표로 모은다
# 왜 하나 — 무엇을 얻고 무엇이 낮아졌는지 한눈에 비교한다

def 성향에_안_맞는_개수(추천함수):
    """300명의 추천을 보고, 그 투자자의 위험도 상한을 넘는 종목 수와 전체 추천 수를 센다."""
    안_맞음 = 0
    전체 = 0
    for 사람 in users["user_id"]:
        추천 = 추천함수(사람, 이미_담은것.get(사람, set()))
        전체 = 전체 + len(추천)
        for 번호 in 추천:
            if 위험도[번호] > 상한[성향[사람]]:                                  # 위험도가 상한보다 크면 센다
                안_맞음 = 안_맞음 + 1
    return 안_맞음, 전체

방식들 = [("1주차", "전체 인기순", 인기순_추천),
         ("2주차", "세그먼트별 인기순", 세그먼트_추천),
         ("4주차", "행렬분해 (상한 없음)", 행렬분해_추천),
         ("6주차", "행렬분해 + 위험도 상한", 상한_추천)]

줄들 = []
for 주차, 이름, 추천함수 in 방식들:
    점수, 채점_인원 = recsys.recall_at_k(추천함수, train, test)                 # 3주차부터 쓰던 채점 함수
    안_맞음, 전체 = 성향에_안_맞는_개수(추천함수)
    시간 = 걸린_시간["4주차 · 6주차"] if 주차 in ("4주차", "6주차") else 걸린_시간[주차]
    줄들.append({"주차": 주차, "방식": 이름, "Recall@10": round(점수, 4),
                 "성향에 맞지 않는 추천": f"{안_맞음} / {전체:,}", "만드는 데 걸린 시간(ms)": round(시간, 1),
                 "_안_맞음": 안_맞음, "_점수": 점수})                                   # _ 로 시작하는 열은 계산용(표시하지 않음)

비교표 = pd.DataFrame(줄들)
print(f"채점 대상 투자자 {채점_인원}명")
비교표.drop(columns=["_안_맞음", "_점수"])
```

**결론: 점수가 가장 높은 것은 4주차(0.3918)이지만, 성향에 맞지 않는 추천이 205개입니다. 6주차는 점수가 0.0019 낮고 그 추천이 0개입니다.**
- **1주차 인기순** — Recall@10 **0.2163** (기준선). 성향에 맞지 않는 추천이 **727개**로 가장 많습니다. 성향을 전혀 보지 않기 때문입니다
- **2주차 세그먼트** — **0.2684**. 취향이 비슷한 무리를 쓰면 점수가 오르지만 성향에 맞지 않는 추천은 **256개**입니다 (이 방식은 일부 투자자에게 10개 미만을 추천해서 전체 추천 수가 2,960개입니다)
- **4주차 행렬분해** — **0.3918**로 가장 높습니다. 그래도 **205개**는 성향에 맞지 않습니다 (6주차에서 센 205개와 같음)
- **6주차 행렬분해 + 상한** — **0.3899**. 성향에 맞지 않는 추천이 **0개**입니다
- 학습 시간은 모두 아주 짧습니다. 실행할 때마다 값이 조금씩 다릅니다. 우리 데이터가 작기 때문입니다

- 점수표에 없는 주차: **5주차**는 후보를 먼저 추려 계산을 줄인 구조입니다(전체 100개 종목 대신 후보 40개만 계산해 계산량 60% 감소, Recall@10 0.3382), **6주차의 FM · MLP** 는 행렬분해보다 낮았습니다(0.3463 · 0.2580). **7주차**는 6주차와 같은 점수에 자동화를, **8 · 9주차**는 점수가 아니라 전달 방법을 더했습니다

### <mark style="display:block; background:#f8bbd0; color:#1a1a1a; padding:6px 12px; border-radius:4px">🎯 10.2 최종 추천 방식 고르기 — 고르는 규칙을 코드로 적용한다</mark>

**흐름에서의 위치:** 비교표 → **고르는 규칙** → 최종 방식 한 개.

**10.2 에서 하는 일** — 「성향에 맞지 않는 추천이 **0개**인 방식 중에서 Recall@10 이 가장 높은 것」이라는 규칙을 표에 적용합니다.

**왜 하나**
- 「점수만 보면 4주차, 그런데 6주차를 고른다」는 선택을 **사람 마음이 아니라 규칙**으로 설명하기 위해서입니다.
- 규칙을 코드로 적으면 누가 해도 같은 결과가 나오고, 데이터가 바뀌어도 같은 규칙을 다시 적용할 수 있습니다.

**순서**
1. 점수만 기준으로 1등을 구한다
2. 적합성 조건(성향에 맞지 않는 추천 0개)을 통과한 것만 남긴다
3. 남은 것 중 점수가 가장 높은 것을 최종으로 고른다
4. 두 선택의 점수 차이(낮아진 점수)를 계산한다

```python
# 이 셀에서 하는 일 — 점수만 본 1등과, 적합성 조건을 통과한 것 중 1등을 비교한다
# 왜 하나 — 고르는 규칙을 적용하고, 그 규칙 때문에 점수가 얼마나 낮아졌는지 숫자로 확인한다

점수_1등 = 비교표.sort_values("_점수", ascending=False).iloc[0]                  # 1) 점수만 기준

통과 = 비교표[비교표["_안_맞음"] == 0]                                           # 2) 성향에 맞지 않는 추천이 0개인 것만

if len(통과) == 0:                                                             # 통과한 방식이 없으면 안내만 한다
    print("적합성 조건을 통과한 방식이 없습니다. 규칙을 다시 정해야 합니다.")
else:
    최종 = 통과.sort_values("_점수", ascending=False).iloc[0]                    # 3) 그중 점수 1등
    점수_차이 = 최종["_점수"] - 점수_1등["_점수"]                                  # 4) 규칙 때문에 바뀐 점수

    print(f"점수만 보면    : {점수_1등['주차']} {점수_1등['방식']} — Recall@10 {점수_1등['_점수']:.4f} · 성향에 맞지 않는 추천 {점수_1등['성향에 맞지 않는 추천']}")
    print(f"규칙을 적용하면 : {최종['주차']} {최종['방식']} — Recall@10 {최종['_점수']:.4f} · 성향에 맞지 않는 추천 {최종['성향에 맞지 않는 추천']}")
    print(f"→ 규칙 때문에 점수 {점수_차이:+.4f}")
```

**결과 읽는 법**

**결론: 규칙을 적용하면 최종은 6주차(행렬분해 + 위험도 상한)입니다. 점수 1등인 4주차보다 0.0019 낮지만 성향에 맞지 않는 추천이 0개입니다.**
- **점수만 보면** 4주차 — 0.3918, 성향에 맞지 않는 추천 205개
- **규칙을 적용하면** 6주차 — 0.3899, 성향에 맞지 않는 추천 0개
- **트레이드오프** — 적합성을 지키려고 점수가 **0.0019** 낮아졌습니다. 6주차에서 센 값과 같습니다
- 이 규칙은 금융의 **적합성 원칙**을 점수보다 앞에 둔 것입니다. 규칙을 바꾸면(예 — 「부적합 추천이 10개 이하이면 허용」) 최종 선택도 바뀔 수 있습니다
- 7주차 파이프라인이 만드는 추천이 바로 이 6주차 방식입니다

```python
# 이 셀에서 하는 일 — 고른 아이디어와 내가 적은 내용으로 프로젝트 계획서 카드를 만든다
# 왜 하나 — 무엇을 · 누구를 위해 · 어떤 기준으로 · 무엇을 조심할지를 한 장으로 정리한다

아이디어 = "2. 신규 투자자 15명에게 투자 성향별 인기순으로 추천하기"   #@param ["1. 안정형 위험도 상한 바꿔 보기", "2. 신규 투자자 15명에게 투자 성향별 인기순으로 추천하기", "3. 행렬분해 요인 수 바꿔 보기", "4. 대시보드에 화면 추가하기", "5. API 에 인증 키 붙이기", "6. 내 업무 데이터에 적용하는 계획"] {allow-input: true}
대상 = "기록이 없어 인기순만 받는 신규 투자자와 상담 직원"   #@param {type:"string"}
성공_기준 = "신규 투자자 15명의 Recall@10 평균이 전체 인기순보다 높아진다"   #@param {type:"string"}
주의할_점 = "신규 투자자는 수가 적어 점수가 크게 흔들릴 수 있다"   #@param {type:"string"}

항목들 = [("아이디어", 아이디어), ("누구를 위해", 대상), ("성공 기준", 성공_기준), ("주의할 점", 주의할_점)]

if any(내용.strip() == "" for 이름, 내용 in 항목들):                                # 비어 있는 항목이 있으면 안내만 한다
    print("네 항목을 모두 적어 주세요. 비어 있는 항목이 있습니다.")
else:
    print("=" * 60)
    print("  프로젝트 계획서")
    print("=" * 60)
    for 이름, 내용 in 항목들:
        print(f"  {이름:8s}: {내용}")
    print("=" * 60)
```

**점검 한 가지 — 결과 파일이 저장소에 올라가는가**
- 9주차 앱은 `output/recommendations.csv` 를 읽습니다. 그런데 이 폴더는 보통 Git 이 **올리지 않도록(무시하도록)** 설정되어 있습니다.
- 그대로 배포하면 앱이 「결과 파일이 없습니다」로 멈춥니다. 배포하려면 결과 파일을 저장소에 함께 올리거나, 앱이 시작할 때 만들게 해야 합니다.

```python
# 이 셀에서 하는 일 — 배포에 필요한 파일이 있는지, 결과 파일이 Git 에서 제외되어 있는지 확인한다
# 왜 하나 — 결과 파일이 저장소에 올라가지 않으면 배포한 앱이 멈추기 때문에, 배포하기 전에 알아낸다

import os                                                                  # 파일이 있는지 확인하는 도구
import subprocess                                                          # 터미널 명령을 파이썬에서 실행하는 도구

필요한_파일 = [("dashboard_app.py", "배포할 앱 코드"),
              ("output/recommendations.csv", "앱이 읽는 결과 파일")]

for 파일_이름, 설명 in 필요한_파일:
    print(f"{'있음' if os.path.exists(파일_이름) else '없음':4s} {파일_이름}  — {설명}")

결과 = subprocess.run(["git", "check-ignore", "-q", "output/recommendations.csv"], capture_output=True, text=True)   # -q = 결과만 알려 주고 글자는 내지 않는다
if 결과.returncode == 0:                                                   # 0 = Git 이 이 파일을 무시한다
    print("\n⚠ output/recommendations.csv 는 Git 이 올리지 않도록 설정되어 있습니다.")
    print("  → 이대로 배포하면 앱이 「결과 파일이 없습니다」로 멈춥니다.")
elif 결과.returncode == 1:                                                 # 1 = 무시하지 않는다
    print("\n결과 파일은 Git 에 올라갑니다.")
else:
    print("\nGit 저장소 안이 아니라서 확인하지 못했습니다.")
```

**결론: 앱 코드와 결과 파일은 있지만, 결과 파일이 Git 에서 제외되어 있어서 그대로 배포하면 앱이 멈춥니다.**
- 두 파일 모두 **있음**입니다 (지금 실행 환경에는 있습니다)
- 그런데 `output/` 폴더가 Git 에서 제외되어 있어서, 배포 저장소에는 **결과 파일이 올라가지 않습니다**
- 해결 방법은 두 가지입니다 — 결과 파일을 저장소에 올리거나(가상 데이터일 때만), 앱이 시작할 때 7주차 스크립트로 파일을 만들게 합니다
